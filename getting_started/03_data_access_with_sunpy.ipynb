````python
from sunpy.net import Fido, attrs as a
````

## Connecting to HPA with `sunpy.net.Fido`

The `Fido` interface provides search- and retrieval capabilities for heliophysical data, including and beyond the HPA. The `search` method accepts `Time`, `Instrument`, `Level`, and other attributes to filter the `Fido` registry.

````python
result = Fido.search(a.Time('2012/3/4', '2012/3/6'), a.Instrument.lyra, a.Level.two)
print(result)
````

````python
result = Fido.search(a.Provider('esa'), a.Time('2012/3/4', '2012/3/6'), a.Instrument.lyra, a.Level.two)
print(result)
````

````python
result = Fido.search(a.Provider('esa'), a.Time('2012/3/4', '2012/3/6'))
````
